```python
# loading the dataset
df = pd.read_csv('ipl_matches.csv')

print("First 5 rows of the dataset:")
print(df.head())

print("\nDataset shape (rows, columns):", df.shape)
```

```python
# checking and handling missing values
print("\nMissing values before cleaning:")
print(df.isnull().sum())

df['player_of_match'] = df['player_of_match'].fillna('Not Announced')

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# adding useful columns using pandas and numpy
df['total_runs'] = df['first_innings_runs'] + df['second_innings_runs']
df['match_type'] = np.where(df['total_runs'] >= 400, 'High Scoring', 'Normal')
```

```python
# 1. Total wins by each team
print("\n--- Total Wins by Each Team ---")
team_wins = df['winner'].value_counts()
print(team_wins)
```

```python
# 2. Team-wise win percentage table
teams = list(set(df['team1']).union(set(df['team2'])))
teams.sort()

summary = []
for t in teams:
    played = len(df[(df['team1'] == t) | (df['team2'] == t)])
    won = len(df[df['winner'] == t])
    lost = played - won
    win_pct = round((won / played) * 100, 2)
    summary.append([t, played, won, lost, win_pct])

team_df = pd.DataFrame(summary, columns=['Team', 'Played', 'Won', 'Lost', 'Win_Percentage'])
team_df = team_df.sort_values(by='Win_Percentage', ascending=False)

print("\n--- Team Performance Summary ---")
print(team_df.to_string(index=False))
```

```python
# 3. Batting first vs Chasing wins
bat_first_wins = len(df[df['win_by_runs'] > 0])
chasing_wins = len(df[df['win_by_wickets'] > 0])

print("\n--- Batting First vs Chasing ---")
print("Matches won by batting first:", bat_first_wins)
print("Matches won by chasing:", chasing_wins)
```

```python
# 4. Toss impact analysis
toss_and_match_win = np.sum(df['toss_winner'] == df['winner'])
toss_win_pct = round((toss_and_match_win / len(df)) * 100, 2)

print("\n--- Toss Impact ---")
print("Toss decisions count:")
print(df['toss_decision'].value_counts())
print("Matches where toss winner also won the match:", toss_and_match_win, "out of", len(df))
print("Toss winner win percentage:", toss_win_pct, "%")
```

```python
# 5. Runs analysis using NumPy
first_inn = np.array(df['first_innings_runs'])
total_match_runs = np.array(df['total_runs'])

print("\n--- Runs Statistics (NumPy) ---")
print("Average 1st innings score:", round(np.mean(first_inn), 2))
print("Highest 1st innings score:", np.max(first_inn))
print("Lowest 1st innings score:", np.min(first_inn))
print("Standard deviation of 1st innings runs:", round(np.std(first_inn), 2))
print("Average total runs per match:", round(np.mean(total_match_runs), 2))

print("\nMatch scoring categories:")
print(df['match_type'].value_counts())
```

```python
# 6. Venue-wise average total runs
print("\n--- Average Total Runs by Venue ---")
venue_avg = df.groupby('venue')['total_runs'].mean().round(2).sort_values(ascending=False)
print(venue_avg)
```

```python
# 7. Biggest victories
max_run_win = df.loc[df['win_by_runs'].idxmax()]
max_wkt_win = df.loc[df['win_by_wickets'].idxmax()]

print("\n--- Biggest Victories ---")
print("Biggest win by runs:", max_run_win['winner'], "won by", max_run_win['win_by_runs'], "runs against",
      max_run_win['team2'] if max_run_win['winner'] == max_run_win['team1'] else max_run_win['team1'])
print("Biggest win by wickets:", max_wkt_win['winner'], "won by", max_wkt_win['win_by_wickets'], "wickets against",
      max_wkt_win['team2'] if max_wkt_win['winner'] == max_wkt_win['team1'] else max_wkt_win['team1'])
```

```python
# 8. Top Players of the Match
print("\n--- Most Player of the Match Awards ---")
print(df['player_of_match'].value_counts().head(5))
```

```python
# 9. Function to check any specific team's record
def check_team(team_name):
    matches = df[(df['team1'] == team_name) | (df['team2'] == team_name)]
    wins = len(matches[matches['winner'] == team_name])
    print("\n--- Quick Report for", team_name, "---")
    print("Total matches played:", len(matches))
    print("Total wins:", wins)
    print("Total losses:", len(matches) - wins)
```

```python
# change team name here to check another team
check_team('RCB')
check_team('GT')
```
